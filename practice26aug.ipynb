```python
#polymorphism

class car():
    def __init__(self,brand):
        self.brand=brand
    def move(self):
        print(f"{self.brand},Drive")
class boat():
    def __init__(self,brand):
        self.brand=brand
    def move(self):
        print(f"{self.brand},Sailing")
class plane():
    def __init__(self,brand):
        self.brand=brand
    def move(self):
        print(f"{self.brand},Fly")                        
Car=car("Ford")
Boat=boat("Ibiza")
Plane=plane("Boeing")
for x in(Car,Boat,Plane):
    x.move()
```

```python
#polymorphism with inheritance
class vehicle():
    def __init__(self,brand):
        self.brand=brand
    def move(self):
        print(f"{self.brand},Move")
class car(vehicle):
    pass
class boat(vehicle):
    def move(self):
        print(f"{self.brand}, Sailing")
class plane(vehicle):
    def __init__(self,brand,model):
        super().__init__(brand)
        self.model=model  
    def move(self):
        print(f"{self.brand} {self.model}, Fly")
Car=car("Ford")
Boat=boat("Ibiza")
Plane=plane("Boieng", "747")
for x in (Car,Boat,Plane):
  x.move()

```

```python
#bubble sort

numbers=[5,6,2,9,1]
n=len(numbers)
for i in range(n):
    for j in range(0,n-i-1):
        if numbers[j]>numbers[j+1]:
            numbers[j], numbers[j+1]=numbers[j+1], numbers[j]
print(numbers)            
```
