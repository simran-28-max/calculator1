```python
#USING SUPER() FUNCTION

class Person:
    def __init__(self, fname, lname):
            self.firstname= fname
            self.lastname= lname
    
    def printname(self):
            print(self.firstname, self.lastname)
class Student(Person):
    def __init__(self, fname, lname):
           super().__init__(fname, lname)  
y=Student("Ekam", "Singh")
y.printname()     
```

```python
#create a program using parent class Person and add the attributes fname, lname and create a printname method
#then create a child class Student, in this class use the super() function too....also add a property graduationyear and method welcome
#then create a object of the child class and execute the welcome method 

class Person:
  def __init__(self, fname, lname):
    self.firstname = fname
    self.lastname = lname

  def printname(self):
    print(self.firstname, self.lastname)

class Student(Person):
  def __init__(self, fname, lname, year):
    super().__init__(fname, lname)
    self.graduationyear = year

  def welcome(self):
    print("Welcome", self.firstname, self.lastname, "to the class of", self.graduationyear)

x = Student("Ekam", "Singh", 2026)
x.welcome()
```
