```python
#find the smallest number in the linked list[4,8,10,12]

class Node:
    def __init__(self,data):
        self.data=data
        self.next=None

node1=Node(4)
node2=Node(8)
node3=Node(10)
node4=Node(12)
       
node1.next=node2
node2.next=node3
node3.next=node4

def minimum_value(head):
    if head is None:
       return None


    #assume minimum value
    minimum_value=head.data
    current=head.next

    while current is not None:
      if current.data<minimum_value:
        minimum_value=current.data
      current=current.next

    return minimum_value 
print("Lowest value:",minimum_value(node1))
```

```python
#stack using class

class Stack:
    def __init__(self):
        self.stack=[]

#add items
    def push(self,element):
     self.stack.append(element)


#remove element

    def pop(self):
       if self.isEmpty():
         return "Stack is empty"
       return self.stack.pop()

# view but not remove

    def peek(self):
       if self.isEmpty:
         return "Stack is empty" 
       return self.stack[-1]

    def isEmpty(self):
       return len(self.stack) == 0

    def size(self):
       return len(self.stack)

myStack=Stack()

myStack.push('A')
myStack.push('B')

print("Stack:", myStack.stack)
print("Pop:",myStack.pop())
print("isEmpty:", myStack.isEmpty())
print("Size:", myStack.size())
```
