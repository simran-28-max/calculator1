```python
import math

principal = 1000  # initial amount
rate = 0.05       # interest rate (5%)
years = 3         # number of years

# Compound interest formula: A = P * (1 + r)^t
amount = principal * math.pow((1 + rate), years)
print(amount)  # Output: 1157.625
```
