```python
import pandas as pd

students={
    "name": ["simran", "karan", "ekam", "simran"],
    "age":  [31,26,None,31],
    "salary": [-5000,40000,45000,-5000],
    "department": ["it", "hr", "hr", "it"]
}

df=pd.DataFrame(students)

#remove duplicates records
df=df.drop_duplicates()

#clean department names
df["department"]=df["department"].str.strip().str.upper()

#correct invalid salaries
df.loc[df["salary"]<0, "salary"]= None

print(df)
```
