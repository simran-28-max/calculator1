```python
import pandas as pd

students={
    "name": ["simran", "karan", "ekam", "simran"],
    "age":  [31,26,None,31],
    "salary": [-5000,40000,45000,-5000],
    "department": ["it", "hr", "hr", "it"]
}

df=pd.DataFrame(students)

#remove duplicates records
df=df.drop_duplicates()

#clean department names
df["department"]=df["department"].str.strip().str.upper()

#correct invalid salaries
df.loc[df["salary"]<0, "salary"]= None

print(df)
```

```python
import pandas as pd
data={
    "name": ["simran", "karan", "aman", "priya", "aman"],
    "age":  [31,26,None, 25, None],
    "dept": ["it", "hr", "admin", None, "admin"], 
    "salary": [40000, 45000, 35000, 40000, 35000]
}
df=pd.DataFrame(data)
print(df)
df.isna().sum()
df.notna().sum()
df.fillna(0)
df["age"]=df["age"].fillna(0)
df.duplicated()
df.drop_duplicates()
df.sort_values("salary")
df.sort_values("salary", ascending=False)
df["bonus"] = df["salary"]*0.10   #add new column
print(df)
df["salary"] = df["salary"]+2500 #update exsiting column
print(df)
df.rename(columns={"salary": "monthly_salary"},
          inplace=True
          )
print(df)
df["monthly_salary"].sum()
df["monthly_salary"].mean()
df["monthly_salary"].max()
df["monthly_salary"].min()
df["monthly_salary"].count()
df["monthly_salary"].median()
df["monthly_salary"].nunique()
df.groupby("dept")["monthly_salary"].mean()
df["monthly_salary"].agg(["sum", "mean", "max", "min"])
```

```python
import matplotlib.pyplot as plt

months=["jan", "feb", "march", "apr"]
sales=[500,300,900,400]

plt.plot(months, sales)
plt.xlabel("Months")
plt.ylabel("Sales")
plt.show()
plt.legend()
plt.grid()
plt.title("Monthly Sales")    
```
