```python
#doubly linked list

class Node:
    def __init__(self,data):
        self.data=data
        self.prev=None
        self.next=None

node1=Node("Ekam")
node2=Node("Simran")
node3=Node("Karan")

node1.next=node2
node2.next=node3
node3.prev=node2
node3.prev.prev=node1

print(node1.data)
print(node1.next.data)
print(node1.next.next.data)
print(node3.prev.prev.data)
print(node3.prev.data)
```

```python
#circular linked list


class Node:
    def __init__(self,data):
        self.data=data
        self.prev=None
        self.next=None

node1=Node("Ekam")
node2=Node("Simran")
node3=Node("Karan")

node1.next=node2
node2.next=node3
node3.next=node1

print(node1.data)
print(node1.next.data)
print(node1.next.next.data)
print(node3.next.data)
```

```python
#traverse of linked list

class Node:
    def __init__(self,data):
        self.data=data
        self.next=None


node1=Node("Ekam")
node2=Node("Simran")
node3=Node("Karan")

node1.next=node2
node2.next=node3


#traverse function

def traverse(head):
    current=head

    while current is not None:
      print(current.data)
      current=current.next#move to the next node

    #run traversal starting from the head node
traverse(node1)
```

```python
#find the lowest value in linked list 10,5,20,3

class Node:
    def __init__(self,data):
        self.data=data
        self.next=None

def find_min(head):
    if head is None:
            return None

    minimum=head.data
    current=head.next

    while current is not None:
        if current.data<minimum:
           minimum=current.data
        current=current.next

    return minimum 


         #creating linked list   
head=Node(10)
head.next=Node(5)
head.next.next=Node(20)
head.next.next.next=Node(3)
print("Lowest value:",find_min(head))
```

```python
#delete a node

class Node:
    def __init__(self,data):
        self.data=data
        self.next=None

node1=Node(7)
node2=Node(11)
node3=Node(3)
node4=Node(2)
node5=Node(9)

node1.next=node2
node2.next=node3
node3.next=node4
node4.next=node5

#traversal method

def traverseAndPrint(head):


    current=head  #pointer variable

    while current is not None:
       print(current.data)
       current=current.next


def deleteSpecificNode(head, nodeToDelete):
    if head==nodeToDelete:
       return head.next   
    
    currentNode=head

    while currentNode.next and currentNode.next!=nodeToDelete:
       currentNode=currentNode.next

    if currentNode.next is None:
      return head

    currentNode.next=currentNode.next.next
    return head

print("Before deletion:")
traverseAndPrint(node1)

node1= deleteSpecificNode(node1,node4)

print("\n After deletion:")
traverseAndPrint(node1)
```
